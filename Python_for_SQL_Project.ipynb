```python
import sqlite3
import numpy as np

# Connect to the SQLite database
conn = sqlite3.connect("south_west_winds.db")
cursor = conn.cursor()

# Query the numeric WindSpeed data from the WindData table
cursor.execute("SELECT WindSpeed FROM WindData")
rows = cursor.fetchall()

# Extract wind speeds into a list
wind_speeds = [row[0] for row in rows]

# Calculate the mean using numpy
mean_wind_speed = np.mean(wind_speeds)

# Calculate expected baseline energy output (1 mph ≈ 20 W)
expected_baseline_energy = mean_wind_speed * 20

# Close database connection
conn.close()

# Output results
print(f"Mean Wind Speed: {mean_wind_speed:.2f} mph")
print(f"Expected Baseline Energy Output: {expected_baseline_energy:.1f} W")
```
